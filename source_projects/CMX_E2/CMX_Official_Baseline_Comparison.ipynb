```python

from pathlib import Path
import sys
import torch

PROJECT = Path("/home/cse-sdpl/Downloads/CMX_Original")
DATASET = PROJECT / "datasets" / "TDD"

print("Python executable:", sys.executable)
print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("GPU memory (GB):",
          round(torch.cuda.get_device_properties(0).total_memory / 1e9, 2))

print("\nProject exists:", PROJECT.exists())
print("Dataset exists:", DATASET.exists())

print("\nExisting trained checkpoint:")
trained = PROJECT / "log_TDD_mit_b2" / "checkpoint" / "epoch-100.pth"
print(trained, "| Exists:", trained.exists())

print("\nPotential pretrained checkpoints:")
for folder_name in ["pretrained", "pretrained_models", "checkpoint", "checkpoints"]:
    folder = PROJECT / folder_name
    if folder.exists():
        for path in folder.rglob("*.pth"):
            print(path.relative_to(PROJECT),
                  f"| {path.stat().st_size / 1e6:.1f} MB")

print("\nDataset split sizes:")
for split in ["train.txt", "val.txt", "test.txt"]:
    path = DATASET / split
    if path.exists():
        count = sum(1 for line in path.open() if line.strip())
        print(f"{split}: {count}")
```

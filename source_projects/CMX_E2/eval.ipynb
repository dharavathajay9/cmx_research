```python
from pathlib import Path
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

PROJECT = Path("/home/cse-sdpl/Downloads/CMX_Original")
DATASET = PROJECT / "datasets" / "TDD"

sample_id = "00498D"

rgb = Image.open(DATASET / "RGB" / f"{sample_id}.png").convert("RGB")
thermal = Image.open(DATASET / "Thermal" / f"{sample_id}.png").convert("L")
ground_truth = np.array(
    Image.open(DATASET / "Label" / f"{sample_id}.png")
)

pred_path = PROJECT / "predictions_TDD_epoch100" / f"{sample_id}.png"

if not pred_path.exists():
    raise FileNotFoundError(
        f"Prediction not found: {pred_path}\n"
        "Check whether Step 6 completed successfully."
    )

prediction = np.array(Image.open(pred_path))

print("Ground-truth classes:", np.unique(ground_truth))
print("Predicted classes:", np.unique(prediction))

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].imshow(rgb)
axes[0, 0].set_title("RGB Input")

axes[0, 1].imshow(thermal, cmap="gray")
axes[0, 1].set_title("Thermal Input")

axes[1, 0].imshow(ground_truth, cmap="nipy_spectral", vmin=0, vmax=8)
axes[1, 0].set_title("Ground Truth — Class IDs")

axes[1, 1].imshow(prediction, cmap="nipy_spectral", vmin=0, vmax=8)
axes[1, 1].set_title("CMX Prediction — Class IDs")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path

DATASET = Path("/home/cse-sdpl/Downloads/CMX_Original/datasets/TDD")

def load_base_ids(filename):
    ids = set()
    for line in (DATASET / filename).read_text().splitlines():
        item = line.strip().split("/")[-1]
        if not item:
            continue
        item = Path(item).stem
        if item.endswith("_flip"):
            item = item[:-5]
        ids.add(item)
    return ids

train = load_base_ids("train.txt")
val = load_base_ids("val.txt")
test = load_base_ids("test.txt")

print("Train unique base IDs:", len(train))
print("Validation unique base IDs:", len(val))
print("Test unique base IDs:", len(test))
print("Train/validation shared base IDs:", sorted(train & val)[:20])
print("Train/test shared base IDs:", sorted(train & test)[:20])
print("Validation/test shared base IDs:", sorted(val & test)[:20])
```

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# Project paths
project = Path("/home/cse-sdpl/Downloads/CMX_Original")
sample = "00498D.png"

rgb_path = project / "datasets/TDD/RGB" / sample
thermal_path = project / "datasets/TDD/Thermal" / sample
gt_path = project / "datasets/TDD/Label" / sample
pred_path = project / "predictions_TDD_epoch100" / sample

# Verify all files before plotting
paths = {
    "RGB": rgb_path,
    "Thermal": thermal_path,
    "Ground Truth": gt_path,
    "CMX Prediction": pred_path,
}

for name, path in paths.items():
    print(f"{name}: {'FOUND' if path.exists() else 'MISSING'} — {path}")

if not all(path.exists() for path in paths.values()):
    raise FileNotFoundError("A required image is missing. Check the paths above.")

# Load images
rgb = Image.open(rgb_path).convert("RGB")
thermal = Image.open(thermal_path).convert("L")
gt = np.array(Image.open(gt_path).convert("L"))
pred = np.array(Image.open(pred_path).convert("L"))

print("\nImage shapes:")
print("RGB:", np.array(rgb).shape)
print("Thermal:", np.array(thermal).shape)
print("Ground truth:", gt.shape, "Classes:", np.unique(gt))
print("Prediction:", pred.shape, "Classes:", np.unique(pred))

# Plot
fig, axes = plt.subplots(1, 4, figsize=(20, 5))

axes[0].imshow(rgb)
axes[0].set_title("RGB Input")

axes[1].imshow(thermal, cmap="gray")
axes[1].set_title("Thermal Input")

axes[2].imshow(gt, vmin=0, vmax=8, cmap="tab10", interpolation="nearest")
axes[2].set_title("Ground Truth")

axes[3].imshow(pred, vmin=0, vmax=8, cmap="tab10", interpolation="nearest")
axes[3].set_title("CMX Prediction — Epoch 100")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from PIL import Image
from pathlib import Path

project = Path("/home/cse-sdpl/Downloads/CMX_Original")

gt = np.array(
    Image.open(project / "datasets/TDD/Label/00498D.png").convert("L")
)
pred = np.array(
    Image.open(project / "predictions_TDD_epoch100/00498D.png").convert("L")
)

class_names = {
    0: "Unlabeled",
    1: "Car",
    2: "Person",
    3: "Bike",
    4: "Curve",
    5: "Car stop",
    6: "Guardrail",
    7: "Color cone",
    8: "Bump",
}

print("SAMPLE-LEVEL SEGMENTATION ANALYSIS")
print("-" * 65)
print(f"{'Class':<15}{'GT pixels':>12}{'Pred pixels':>14}{'IoU':>12}")

for class_id, name in class_names.items():
    gt_mask = gt == class_id
    pred_mask = pred == class_id

    gt_count = int(gt_mask.sum())
    pred_count = int(pred_mask.sum())
    union = int((gt_mask | pred_mask).sum())
    intersection = int((gt_mask & pred_mask).sum())

    if union > 0:
        iou = intersection / union
        iou_text = f"{iou * 100:.2f}%"
    else:
        iou_text = "N/A"

    print(f"{name:<15}{gt_count:>12}{pred_count:>14}{iou_text:>12}")

print("-" * 65)
print("Pixel accuracy:", f"{(gt == pred).mean() * 100:.2f}%")
```

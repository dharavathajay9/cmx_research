```python
from pathlib import Path

DATASET = Path("/home/cse-sdpl/Downloads/CMX_Original/datasets/TDD")

def load_base_ids(filename):
    ids = set()
    for line in (DATASET / filename).read_text().splitlines():
        item = line.strip().split("/")[-1]
        if not item:
            continue
        item = Path(item).stem
        if item.endswith("_flip"):
            item = item[:-5]
        ids.add(item)
    return ids

train = load_base_ids("train.txt")
val = load_base_ids("val.txt")
test = load_base_ids("test.txt")

print("Train unique base IDs:", len(train))
print("Validation unique base IDs:", len(val))
print("Test unique base IDs:", len(test))
print("Train/validation shared base IDs:", sorted(train & val)[:20])
print("Train/test shared base IDs:", sorted(train & test)[:20])
print("Validation/test shared base IDs:", sorted(val & test)[:20])
```

```python

from pathlib import Path
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path("/home/cse-sdpl/Downloads/CMX_E1")
PRED_DIR = ROOT / "e1_predictions"
GT_DIR = ROOT / "datasets/TDD/Label"
VAL_FILE = ROOT / "datasets/TDD/val.txt"

classes = [
    "unlabeled", "car", "person", "bike", "curve",
    "car_stop", "guardrail", "color_cone", "bump"
]
NUM_CLASSES = len(classes)

assert PRED_DIR.is_dir(), f"Predictions folder missing: {PRED_DIR}"
assert GT_DIR.is_dir(), f"Ground-truth folder missing: {GT_DIR}"
assert VAL_FILE.is_file(), f"Validation list missing: {VAL_FILE}"

# Read validation image IDs
image_ids = [
    line.strip().split()[0]
    for line in VAL_FILE.read_text().splitlines()
    if line.strip()
]

cm = np.zeros((NUM_CLASSES, NUM_CLASSES), dtype=np.int64)
processed = 0
missing = []

for image_id in image_ids:
    name = Path(image_id).stem
    pred_path = PRED_DIR / f"{name}.png"
    gt_path = GT_DIR / f"{name}.png"

    if not pred_path.exists() or not gt_path.exists():
        missing.append((name, pred_path.exists(), gt_path.exists()))
        continue

    pred = cv2.imread(str(pred_path), cv2.IMREAD_UNCHANGED)
    gt = cv2.imread(str(gt_path), cv2.IMREAD_UNCHANGED)

    if pred is None or gt is None:
        missing.append((name, "read error", "read error"))
        continue

    if pred.ndim == 3:
        pred = pred[:, :, 0]
    if gt.ndim == 3:
        gt = gt[:, :, 0]

    if pred.shape != gt.shape:
        pred = cv2.resize(
            pred, (gt.shape[1], gt.shape[0]),
            interpolation=cv2.INTER_NEAREST
        )

    pred = pred.astype(np.int64)
    gt = gt.astype(np.int64)

    valid = (
        (gt >= 0) & (gt < NUM_CLASSES) &
        (pred >= 0) & (pred < NUM_CLASSES)
    )

    encoded = NUM_CLASSES * gt[valid] + pred[valid]
    cm += np.bincount(
        encoded, minlength=NUM_CLASSES**2
    ).reshape(NUM_CLASSES, NUM_CLASSES)

    processed += 1

print(f"Processed images: {processed}/{len(image_ids)}")
print(f"Missing/unreadable pairs: {len(missing)}")

if missing:
    print("First missing pairs:", missing[:10])

# Confusion matrix: rows = ground truth, columns = predictions
cm_df = pd.DataFrame(
    cm, index=classes, columns=classes
)
display(cm_df)

# Calculate metrics from the confusion matrix
tp = np.diag(cm).astype(float)
gt_count = cm.sum(axis=1).astype(float)
pred_count = cm.sum(axis=0).astype(float)
union = gt_count + pred_count - tp

iou = np.divide(
    tp, union,
    out=np.zeros_like(tp),
    where=union != 0
)
class_acc = np.divide(
    tp, gt_count,
    out=np.zeros_like(tp),
    where=gt_count != 0
)

print(f"Independent mIoU: {iou.mean() * 100:.3f}%")
print(f"Pixel accuracy: {tp.sum() / max(cm.sum(), 1) * 100:.3f}%")

metrics_df = pd.DataFrame({
    "Class": classes,
    "IoU (%)": iou * 100,
    "Class accuracy (%)": class_acc * 100,
    "Ground-truth pixels": gt_count.astype(np.int64)
})
display(metrics_df.round(3))

# Plot the confusion matrix
row_totals = cm.sum(axis=1, keepdims=True)
cm_normalized = np.divide(
    cm, row_totals,
    out=np.zeros_like(cm, dtype=float),
    where=row_totals != 0
)

fig, ax = plt.subplots(figsize=(11, 9))
im = ax.imshow(cm_normalized, cmap="Blues", vmin=0, vmax=1)
ax.set_xticks(range(NUM_CLASSES), classes, rotation=45, ha="right")
ax.set_yticks(range(NUM_CLASSES), classes)
ax.set_xlabel("Predicted class")
ax.set_ylabel("Ground-truth class")
ax.set_title("CMX E1 — Normalized Confusion Matrix")
fig.colorbar(im, ax=ax, label="Fraction of ground-truth pixels")

for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        ax.text(
            j, i, f"{cm_normalized[i, j]:.2f}",
            ha="center", va="center",
            color="white" if cm_normalized[i, j] > 0.55 else "black",
            fontsize=8
        )

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path

CLASS_NAMES = [
    "unlabeled", "car", "person", "bike", "curve",
    "car_stop", "guardrail", "color_cone", "bump"
]
N = len(CLASS_NAMES)

E0_PRED = Path("/home/cse-sdpl/Downloads/CMX_Original/predictions_TDD_epoch100")
E1_PRED = Path("/home/cse-sdpl/Downloads/CMX_E1/e1_predictions")
GT_DIR = Path("/home/cse-sdpl/Downloads/CMX_Original/datasets/TDD/Label")
VAL_FILE = Path("/home/cse-sdpl/Downloads/CMX_Original/datasets/TDD/val.txt")

def get_names():
    return [Path(x.strip()).stem
            for x in VAL_FILE.read_text().splitlines() if x.strip()]

def build_cm(pred_dir):
    cm = np.zeros((N, N), dtype=np.int64)
    missing = []
    ignored_gt_pixels = 0
    invalid_pred_pixels = 0

    for name in get_names():
        gt_path = GT_DIR / f"{name}.png"
        pred_path = pred_dir / f"{name}.png"

        if not gt_path.exists() or not pred_path.exists():
            missing.append(name)
            continue

        gt = np.asarray(Image.open(gt_path))
        pred = np.asarray(Image.open(pred_path))

        if gt.ndim == 3:
            gt = gt[..., 0]
        if pred.ndim == 3:
            pred = pred[..., 0]

        if gt.shape != pred.shape:
            raise ValueError(
                f"{name}: GT shape {gt.shape} != prediction shape {pred.shape}"
            )

        # Ground-truth labels outside 0..8 are excluded (e.g. ignore label 255).
        valid_gt = (gt >= 0) & (gt < N)
        ignored_gt_pixels += int((~valid_gt).sum())

        invalid_pred_pixels += int(
            (valid_gt & ((pred < 0) | (pred >= N))).sum()
        )
        valid = valid_gt & (pred >= 0) & (pred < N)

        encoded = N * gt[valid].astype(np.int64) + pred[valid].astype(np.int64)
        cm += np.bincount(encoded, minlength=N*N).reshape(N, N)

    return cm, missing, ignored_gt_pixels, invalid_pred_pixels

cm_e0, miss_e0, ignored_e0, invalid_e0 = build_cm(E0_PRED)
cm_e1, miss_e1, ignored_e1, invalid_e1 = build_cm(E1_PRED)

print("E0 missing masks:", len(miss_e0))
print("E1 missing masks:", len(miss_e1))
print("E0 ignored GT pixels:", ignored_e0, "| invalid predictions:", invalid_e0)
print("E1 ignored GT pixels:", ignored_e1, "| invalid predictions:", invalid_e1)

if miss_e0 or miss_e1:
    raise RuntimeError("Some masks are missing. Check the printed counts.")

def calculate_metrics(cm):
    tp = np.diag(cm).astype(float)
    gt_count = cm.sum(axis=1).astype(float)
    pred_count = cm.sum(axis=0).astype(float)
    union = gt_count + pred_count - tp

    iou = np.divide(tp, union, out=np.full(N, np.nan), where=union > 0)
    dice = np.divide(
        2 * tp, gt_count + pred_count,
        out=np.full(N, np.nan), where=(gt_count + pred_count) > 0
    )
    class_acc = np.divide(
        tp, gt_count, out=np.full(N, np.nan), where=gt_count > 0
    )

    total = cm.sum()
    pixel_acc = tp.sum() / total if total else np.nan

    return {
        "iou": iou * 100,
        "dice": dice * 100,
        "class_acc": class_acc * 100,
        "miou": np.nanmean(iou) * 100,
        "pixel_acc": pixel_acc * 100
    }

m0 = calculate_metrics(cm_e0)
m1 = calculate_metrics(cm_e1)

comparison = pd.DataFrame({
    "Class": CLASS_NAMES,
    "E0 IoU (%)": m0["iou"],
    "E1 IoU (%)": m1["iou"],
    "Change (percentage points)": m1["iou"] - m0["iou"],
    "E0 Dice (%)": m0["dice"],
    "E1 Dice (%)": m1["dice"]
})

display(comparison.round(3))

summary = pd.DataFrame({
    "Metric": ["mIoU (%)", "Pixel accuracy (%)"],
    "E0 confusion matrix": [m0["miou"], m0["pixel_acc"]],
    "E1 confusion matrix": [m1["miou"], m1["pixel_acc"]],
    "Change (pp)": [
        m1["miou"] - m0["miou"],
        m1["pixel_acc"] - m0["pixel_acc"]
    ]
})
display(summary.round(3))

def plot_cm(cm, title, normalize=False):
    values = cm.astype(float)

    if normalize:
        row_sum = values.sum(axis=1, keepdims=True)
        values = np.divide(
            values, row_sum, out=np.zeros_like(values), where=row_sum != 0
        )

    fig, ax = plt.subplots(figsize=(10, 8))
    im = ax.imshow(values, cmap="Blues", interpolation="nearest")
    fig.colorbar(im, ax=ax)

    ax.set_xticks(range(N), CLASS_NAMES, rotation=45, ha="right")
    ax.set_yticks(range(N), CLASS_NAMES)
    ax.set_xlabel("Predicted class")
    ax.set_ylabel("Ground-truth class")
    ax.set_title(title)

    for i in range(N):
        for j in range(N):
            label = f"{values[i,j]:.2f}" if normalize else f"{int(cm[i,j]):,}"
            ax.text(i*0+j, i, label, ha="center", va="center",
                    color="white" if values[i,j] > values.max()*0.55 else "black",
                    fontsize=8)

    plt.tight_layout()
    plt.show()

plot_cm(cm_e0, "CMX Original (E0) — Confusion Matrix")
plot_cm(cm_e1, "CMX E1 — Confusion Matrix")
plot_cm(cm_e0, "CMX Original (E0) — Row-normalized", normalize=True)
plot_cm(cm_e1, "CMX E1 — Row-normalized", normalize=True)

comparison.to_csv("cmx_e0_e1_class_comparison.csv", index=False)
np.savez("cmx_e0_e1_confusion_matrices.npz", cm_e0=cm_e0, cm_e1=cm_e1)

print("Saved comparison CSV and confusion matrices.")
print("Logged evaluation targets: E0 mIoU 62.726%; E1 mIoU 60.944%.")
```

```python

from pathlib import Path
import shutil

experiments = {
    "E2": Path("/home/cse-sdpl/Downloads/CMX_E2"),
    "E3": Path("/home/cse-sdpl/Downloads/CMX_E3"),
}

def patch_test_support(name, root):
    config_file = root / "config.py"
    eval_file = root / "eval.py"
    loader_file = root / "dataloader" / "RGBXDataset.py"

    cfg = config_file.read_text()
    ev = eval_file.read_text()
    loader = loader_file.read_text()

    # 1. Add test.txt path to config.py
    if "C.test_source" not in cfg:
        marker = "C.eval_source = osp.join(C.dataset_path, 'val.txt')"
        if marker not in cfg:
            raise RuntimeError(f"{name}: cannot locate eval_source in config.py")
        cfg = cfg.replace(
            marker,
            marker + "\nC.test_source = osp.join(C.dataset_path, 'test.txt')",
            1
        )

    # 2. Add split argument
    if "--split" not in ev:
        marker = "    args = parser.parse_args()"
        if marker not in ev:
            raise RuntimeError(f"{name}: cannot locate argument parser")
        ev = ev.replace(
            marker,
            "    parser.add_argument('--split', "
            "choices=['train', 'val', 'test'], default='val')\n" + marker,
            1
        )

    # 3. Add test source to the data_setting dictionary
    if "'test_source': config.test_source" not in ev:
        marker = "                    'eval_source': config.eval_source,"
        if marker not in ev:
            raise RuntimeError(f"{name}: cannot locate eval_source in eval.py")
        ev = ev.replace(
            marker,
            marker + "\n                    'test_source': config.test_source,",
            1
        )

    # 4. Use the selected split
    ev = ev.replace(
        "dataset = RGBXDataset(data_setting, 'val', val_pre)",
        "dataset = RGBXDataset(data_setting, args.split, val_pre)",
        1
    )

    if "RGBXDataset(data_setting, args.split, val_pre)" not in ev:
        raise RuntimeError(f"{name}: split selection was not configured")

    # 5. Add test source to dataset loader
    if "self._test_source" not in loader:
        marker = "        self._eval_source = setting['eval_source']"
        if marker not in loader:
            raise RuntimeError(f"{name}: cannot locate loader eval_source")
        loader = loader.replace(
            marker,
            marker + "\n        self._test_source = setting['test_source']",
            1
        )

    if "assert split_name in ['train', 'val', 'test']" not in loader:
        marker = "        assert split_name in ['train', 'val']"
        if marker not in loader:
            raise RuntimeError(f"{name}: cannot locate split assertion")
        loader = loader.replace(
            marker,
            "        assert split_name in ['train', 'val', 'test']",
            1
        )

    if 'elif split_name == "test":' not in loader:
        marker = '        if split_name == "val":\n            source = self._eval_source'
        if marker not in loader:
            raise RuntimeError(f"{name}: cannot locate split-selection block")
        loader = loader.replace(
            marker,
            marker + '\n        elif split_name == "test":\n'
                     '            source = self._test_source',
            1
        )

    # Validate all modified files before writing
    compile(cfg, str(config_file), "exec")
    compile(ev, str(eval_file), "exec")
    compile(loader, str(loader_file), "exec")

    # Back up current versions once
    for path in (config_file, eval_file, loader_file):
        backup = path.with_suffix(path.suffix + ".repair.bak")
        if not backup.exists():
            shutil.copy2(path, backup)

    config_file.write_text(cfg)
    eval_file.write_text(ev)
    loader_file.write_text(loader)

    print(f"{name}: patched; syntax checks passed.")

for name, root in experiments.items():
    patch_test_support(name, root)

print("\nE2 and E3 are ready for verification, bro!")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# Save all research results in one folder
OUT = Path("/home/cse-sdpl/Downloads/CMX_Research_Matrices")
OUT.mkdir(parents=True, exist_ok=True)

experiments = ["E0", "E1", "E2", "E3"]
splits = ["train", "validation", "test"]

# Overall evaluation metrics (%)
metrics = {
    "train": {
        "E0": [92.720, 99.535, 96.772, 99.092],
        "E1": [93.043, 99.557, 96.913, 99.134],
        "E2": [92.646, 99.528, 96.752, 99.080],
        "E3": [94.915, 99.670, 97.607, 99.351],
    },
    "validation": {
        "E0": [62.726, 97.732, 72.982, 95.784],
        "E1": [60.944, 97.684, 70.880, 95.678],
        "E2": [62.325, 97.731, 72.627, 95.773],
        "E3": [61.961, 97.723, 71.182, 95.739],
    },
    "test": {
        "E0": [58.923, 98.227, 70.841, 96.777],
        "E1": [56.786, 98.176, 65.979, 96.664],
        "E2": [57.830, 98.224, 67.892, 96.768],
        "E3": [57.894, 98.247, 67.030, 96.792],
    },
}

metric_names = [
    "mIoU (%)",
    "Pixel Accuracy (%)",
    "Mean Pixel Accuracy (%)",
    "Frequency Weighted IoU (%)"
]

# M1: Overall metric matrix for each split
overall_tables = {}

for split in splits:
    df = pd.DataFrame(
        metrics[split],
        index=metric_names
    ).T
    overall_tables[split] = df
    df.to_csv(OUT / f"M1_{split}_metrics.csv")

print("M1 — TRAIN METRICS")
display(overall_tables["train"])

print("M1 — VALIDATION METRICS")
display(overall_tables["validation"])

print("M1 — TEST METRICS")
display(overall_tables["test"])

# M2: mIoU across train, validation and test
miou_split_matrix = pd.DataFrame({
    split: {
        exp: metrics[split][exp][0]
        for exp in experiments
    }
    for split in splits
})
miou_split_matrix.to_csv(OUT / "M2_splitwise_mIoU.csv")

print("M2 — SPLIT-WISE mIoU")
display(miou_split_matrix)

# M3: Per-class IoU values (%)
classes = [
    "Unlabeled", "Car", "Person", "Bike", "Curve",
    "Car stop", "Guardrail", "Color cone", "Bump"
]

class_iou = {
    "train": {
        "E0": [99.520,97.232,89.233,90.006,90.455,92.457,95.618,85.688,94.267],
        "E1": [99.543,97.360,89.765,90.399,90.996,92.626,95.882,86.249,94.563],
        "E2": [99.514,97.196,89.110,89.782,90.328,92.313,95.834,85.626,94.115],
        "E3": [99.659,97.970,91.854,92.752,93.865,94.882,97.351,89.706,96.198],
    },
    "test": {
        "E0": [98.303,89.098,75.033,64.802,47.548,33.639,9.273,51.472,61.139],
        "E1": [98.237,88.676,74.412,64.871,46.047,29.591,6.372,44.308,58.564],
        "E2": [98.296,89.336,75.209,64.498,49.209,29.546,6.311,52.058,56.012],
        "E3": [98.314,89.556,75.183,64.706,47.586,30.836,6.209,51.353,57.304],
    }
}

for split in ["train", "test"]:
    df = pd.DataFrame(
        {exp: class_iou[split][exp] for exp in experiments},
        index=classes
    )
    df.index.name = "Class"
    df.to_csv(OUT / f"M3_{split}_classwise_IoU.csv")

    print(f"M3 — {split.upper()} CLASS-WISE IoU")
    display(df.round(3))

# M4: Improvement / difference relative to E0, in percentage points
improvement = pd.DataFrame(index=experiments, columns=metric_names, dtype=float)

for exp in experiments:
    for j, metric in enumerate(metric_names):
        improvement.loc[exp, metric] = (
            metrics["test"][exp][j] - metrics["test"]["E0"][j]
        )

improvement.to_csv(OUT / "M4_test_improvement_vs_E0.csv")

print("M4 — TEST METRIC CHANGE VS E0 (percentage points)")
display(improvement.round(3))

# M5: Generalization gaps (train score minus validation/test score)
gap_rows = []

for exp in experiments:
    train_miou = metrics["train"][exp][0]
    val_miou = metrics["validation"][exp][0]
    test_miou = metrics["test"][exp][0]

    gap_rows.append({
        "Experiment": exp,
        "Train mIoU (%)": train_miou,
        "Validation mIoU (%)": val_miou,
        "Test mIoU (%)": test_miou,
        "Train - Validation (pp)": train_miou - val_miou,
        "Train - Test (pp)": train_miou - test_miou,
        "Validation - Test (pp)": val_miou - test_miou,
    })

generalization = pd.DataFrame(gap_rows).set_index("Experiment")
generalization.to_csv(OUT / "M5_generalization_gaps.csv")

print("M5 — GENERALIZATION GAP MATRIX")
display(generalization.round(3))

print(f"\nAll available matrices saved in: {OUT}")
```

```python
import os
import subprocess
import torch

PID = 306465

print("=== 1. CUDA availability in this notebook ===")
print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

print("\n=== 2. GPU device files opened by E5 ===")
fd_dir = f"/proc/{PID}/fd"
gpu_fds = []

try:
    for fd in os.listdir(fd_dir):
        try:
            target = os.readlink(f"{fd_dir}/{fd}")
            if "nvidia" in target.lower():
                gpu_fds.append((fd, target))
        except (OSError, PermissionError):
            pass

    print(gpu_fds if gpu_fds else
          "No NVIDIA device files found in E5 process descriptors.")
except (OSError, PermissionError) as e:
    print("Could not inspect E5 descriptors:", e)

print("\n=== 3. Processes using NVIDIA device files ===")
result = subprocess.run(
    ["bash", "-lc", "fuser -v /dev/nvidia* 2>&1"],
    capture_output=True, text=True
)
print(result.stdout or result.stderr or "No output")

print("\n=== 4. Is E5 still alive? ===")
result = subprocess.run(
    ["ps", "-p", str(PID), "-o", "pid,etime,%cpu,%mem,stat,args"],
    capture_output=True, text=True
)
print(result.stdout or "E5 process is no longer running.")
```

```python
import os
import signal
import time
from pathlib import Path

PID = 306465

try:
    os.kill(PID, signal.SIGINT)
    print(f"Sent graceful interrupt to E5 process {PID}.")
except ProcessLookupError:
    print("E5 process has already exited.")

# Wait up to 15 seconds for shutdown
for _ in range(30):
    try:
        os.kill(PID, 0)
        time.sleep(0.5)
    except ProcessLookupError:
        break

try:
    os.kill(PID, 0)
    print("E5 is still running. Do not launch another run yet.")
except ProcessLookupError:
    print("E5 has stopped successfully.")

print("\nCheckpoint files:")
ckpt = Path(
    "/home/cse-sdpl/Downloads/CMX_E5/log_TDD_mit_b2_e5/checkpoint"
)
print([p.name for p in ckpt.glob("*.pth")])
```

```python
from pathlib import Path

root = Path("/home/cse-sdpl/Downloads/CMX_E5")

for filename in ["train.py", "engine/engine.py"]:
    path = root / filename
    lines = path.read_text().splitlines()

    print(f"\n========== {filename} ==========")

    keywords = [
        ".cuda(",
        ".to(device",
        "torch.device",
        "config.device",
        "cuda:",
        "set_device",
        "DataParallel",
        "model = model",
    ]

    shown = set()

    for i, line in enumerate(lines):
        if any(k in line for k in keywords):
            for j in range(max(0, i - 2), min(len(lines), i + 3)):
                if j not in shown:
                    print(f"{j + 1}: {lines[j]}")
                    shown.add(j)
            print()
```

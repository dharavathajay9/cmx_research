```python
from pathlib import Path

project = Path("/home/cse-sdpl/Downloads/CMX_E5")

for filename in ["train.py", "config.py"]:
    path = project / filename
    print(f"\n{'='*25} {filename} {'='*25}")

    lines = path.read_text(errors="ignore").splitlines()

    for i, line in enumerate(lines):
        if any(key in line.lower() for key in [
            "resume", "epoch-last", "epoch_last",
            "pretrained", "epoch-100", "mit_b2_e4",
            "checkpoint", "start_epoch"
        ]):
            start = max(0, i - 2)
            end = min(len(lines), i + 3)
            print(f"\nLines {start+1}-{end}:")
            for j in range(start, end):
                print(f"{j+1}: {lines[j]}")
```

```python
import subprocess
from pathlib import Path

root = str(Path("/home/cse-sdpl/Downloads/CMX_E5"))

result = subprocess.run(
    ["ps", "-eo", "pid,args"],
    capture_output=True,
    text=True
)

killed = []

for line in result.stdout.splitlines():
    if root in line and "train.py" in line and "ps -eo" not in line:
        parts = line.strip().split(None, 1)
        if parts and parts[0].isdigit():
            pid = int(parts[0])
            subprocess.run(["kill", "-TERM", str(pid)])
            killed.append(pid)

print("Stop signal sent to E5 process IDs:", killed)
```
